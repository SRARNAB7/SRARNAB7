```python
def cms_ages(age):
  ranges = [
        {'min': 0,  'max': 18,  'label': 'A.  0-18'},
        {'min': 19, 'max': 44,  'label': 'B. 19-44'},
        {'min': 45, 'max': 64,  'label': 'C. 45-64'},
        {'min': 65, 'max': 84,  'label': 'D. 65-84'},
        {'min': 85, 'max': 999, 'label': 'E. 85-up'}
    ]

  for r in ranges:
    if age>= r.get('min') and age <= r.get('max'):
      return r.get('label')

  return 'Z.unknown'
```

Write a Python Program to Check Prime Number.

```python
num = int(input("Enter the number"))

flag = False
if (num==1):
  print(f"{num}, is not a prime number")
elif num > 1:
  for i in range(2,num):
    if(num%i==0):
      flag = True
      break

if flag == True:
  print("It is not a prime number")
else:
  print("It is a prime number")


```

Write a Python program to find largest number in a list.

```python
numbers = [30, 10, -45, 5, 20]

def largest_num(numbers):

  large = int()
  for num in  numbers:
    #large = int()
    if num > large:
      large = num

  print(f'The largest number is {large}')
```

```python
numbers = [30, 10, -45, 5, 10]
largest_num(numbers)
```

Write a Python program to find second largest number in a list.

```python
def second_largest(numbers):

  numbers.sort(reverse = True)
  second_largest = numbers[1]

  print(second_largest)
```

```python
numbers = [23,5,66,87,91]
second_largest(numbers)
```

Write a Python program to print even numbers in a list.

```python
def even_numbers(numbers):
  num = []

  for i in numbers:
    if i%2 == 0:
      num.append(i)

  print(num)
```

```python
numbers = [2,6,9,4,0,1]
even_numbers(numbers)
```

Write a Python program to Count occurrences of an element in a list.

```python
my_list = [1, 2, 3, 4, 2, 5, 2, 3, 4, 6, 5]

def occurences(my_list):

  count_occur = {}

  for num in my_list:
    if num in count_occur:
      count_occur[num] +=1
    else:
      count_occur[num] = 1

  print(count_occur)
```

```python
my_list = [1, 2, 3, 4, 2, 5, 2, 3, 4, 6, 5]
occurences(my_list)
```
